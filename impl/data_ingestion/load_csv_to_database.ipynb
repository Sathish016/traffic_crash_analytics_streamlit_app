```python
import pandas as pd
import mysql.connector

path = r"Testdata/Traffic_CrashesData.csv"
df = pd.read_csv(path)
#df.info()

connection = mysql.connector.connect(
    host="0.0.0.0",
    user="XXXX",
    password="XXXXXXX",
    database="XXXXXX"
)

if connection.is_connected():
    print("Successfully connected to the database")

cursor = connection.cursor()

insert_query = """
INSERT INTO Chicago_Traffic_Crashes (
    CRASH_RECORD_ID,
    CRASH_DATE,
    POSTED_SPEED_LIMIT,
    TRAFFIC_CONTROL_DEVICE,
    WEATHER_CONDITION,
    LIGHTING_CONDITION,
    FIRST_CRASH_TYPE,
    TRAFFICWAY_TYPE,
    PRIM_CONTRIBUTORY_CAUSE,
    SEC_CONTRIBUTORY_CAUSE,
    INJURIES_FATAL,
    INJURIES_INCAPACITATING,
    INJURIES_NON_INCAPACITATING,
    INJURIES_TOTAL,
    CRASH_HOUR,
    CRASH_DAY_OF_WEEK,
    CRASH_MONTH,
    BEAT_OF_OCCURRENCE,
    STREET_NAME,
    LATITUDE,
    LONGITUDE,
    DATE_POLICE_NOTIFIED,
    YEAR
) VALUES (
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s
)
"""

for _, row in df.iterrows():
    values = (
        row.get("CRASH_RECORD_ID"),
        row.get("CRASH_DATE"),
        row.get("POSTED_SPEED_LIMIT"),
        row.get("TRAFFIC_CONTROL_DEVICE"),
        row.get("WEATHER_CONDITION"),
        row.get("LIGHTING_CONDITION"),
        row.get("FIRST_CRASH_TYPE"),
        row.get("TRAFFICWAY_TYPE"),
        row.get("PRIM_CONTRIBUTORY_CAUSE"),
        row.get("SEC_CONTRIBUTORY_CAUSE"),
        row.get("INJURIES_FATAL"),
        row.get("INJURIES_INCAPACITATING"),
        row.get("INJURIES_NON_INCAPACITATING"),
        row.get("INJURIES_TOTAL"),
        row.get("CRASH_HOUR"),
        row.get("CRASH_DAY_OF_WEEK"),
        row.get("CRASH_MONTH"),
        row.get("BEAT_OF_OCCURRENCE"),
        row.get("STREET_NAME"),
        row.get("LATITUDE"),
        row.get("LONGITUDE"),
        row.get("DATE_POLICE_NOTIFIED"),
        row.get("year")
    )
    cursor.execute(insert_query, values)

connection.commit()

print(f"{cursor.rowcount} records inserted.")

cursor.close()
connection.close()
```
